```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Name': ['A','B','C','D','E'],
    'Math' : [78,67,90,56,88],
    'Science' : [85,72,95,60,84],
    'English' : [88,70,92,58,86]
}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Math", "Science", "English"]].values
average_marks = np.mean(marks, axis=1)
df["Average Marks"] = average_marks
print(df)
```

```python
def grade(avg):
  if avg >= 90:
    return 'A'
  elif avg >= 80:
    return 'B'
  elif avg >= 70:
    return 'C'
  elif avg >= 60:
    return 'D'
  else:
    return 'F'
```

```python
# Average marks per subject

df[["Math", "Science", "English"]].mean().plot(kind="bar")
plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.show()
```
