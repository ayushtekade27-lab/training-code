```python
# Writing data to .CSV file

import csv

with open("emp.csv", "w", newline="") as f:
    w = csv.writer(f)

    w.writerow(["ENO", "ENAME", "ESALARY", "EADDR"])

    n = int(input("Enter number of employees : "))

    for i in range(n):
        eno = input("Enter employee no : ")
        ename = input("Enter employee name : ")
        esal = input("Enter employee salary : ")
        eaddr = input("Enter employee address : ")

        w.writerow([eno, ename, esal, eaddr])

print("Data inserted successfully!")
```

```python
import numpy as np

student = np.array([
    [85, 80, 70],  # Student 1
    [80, 55, 70],  # Student 2
    [85, 92, 88],  # Student 3
    [80, 90, 45],  # Student 4
    [72, 68, 75]   # Student 5
])

student
```

```python
avg_marks = np.mean(student, axis=1)
print(avg_marks)
```

```python
topper = np.argmax(avg_marks)
print(topper)
```

```python
failed_students = np.mean(avg_marks < 50)
print(failed_students)
```

```python
for i, avg in enumerate(avg_marks):
    status = "pass" if avg > 50 else "fail"
    print(f"Student {i + 1}: Average Marks = {avg:.2f} ({status})")
```

```python
x = [1,2,3,4]
y = [2,4,6,8]

plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
#bar chat
months = ['jan','feb','march','april',]
sales = [200,300,400,500]
plt.bar(months,sales)
plt.title("Bar Chart")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.barh(months,sales)
plt.title("monthly sales ")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#scatter plot height vs weight
height = [150,160,170,180,190]
weight = [50,60,70,80,90]
plt.scatter(height,weight,color='r')
plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```

```python
#pie chart ( market share)
brands = ['A','B','C','D']
market_share = [20,30,40,10]
plt.pie(market_share, labels=brands, autopct='%1.1f%%')
plt.show()
```
